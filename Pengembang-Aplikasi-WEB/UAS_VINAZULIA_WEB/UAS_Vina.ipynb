```python
from pyspark import SparkContext

# Inisialisasi Spark
try:
    sc = SparkContext.getOrCreate()  # Use existing context if available
except Exception as e:
    print(f"Error initializing SparkContext: {e}")
    exit(1)

# Data nama dan nilai
data = [("Andi", 90), ("Budi", 75), ("Citra", 85), ("Dewi", 95), ("Eka", 70)]

# Membuat RDD
rdd = sc.parallelize(data)

# a. Jumlah total elemen
jumlah = rdd.count()
print(f"Jumlah total elemen: {jumlah}")

# b. Rata-rata nilai
total_nilai = rdd.map(lambda x: x[1]).reduce(lambda a, b: a + b)
rata_rata = total_nilai / jumlah if jumlah > 0 else 0  # Menghindari pembagian dengan nol
print(f"Rata-rata nilai: {rata_rata:.2f}")

# c. Data dengan nilai di atas 80
di_atas_80 = rdd.filter(lambda x: x[1] > 80).collect()
print("Data dengan nilai di atas 80:")
for nama, nilai in di_atas_80:
    print(f"{nama}: {nilai}")

# Stop Spark
sc.stop()
```
