```python
import pandas as pd
```

```python
df = pd.read_csv('data_cleaned.csv')

print('Dataset:', df.shape)
print('\nData Info:', df.info())
df.head()
```

```python
df.drop(columns=['timestamp', 'email_address'], axis=1, inplace=True, errors='ignore')

df
```

```python
# Konversi nilai IPK dan kehadiran
df['ipk'] = df['ipk'].astype(str).str.replace(',', '.').astype(float)
df['kehadiran'] = df['kehadiran'].astype(str).str.replace('%', '').astype(float)

# Hitung mata kuliah favorit tiap semester
def count_favorites(column):
    return df[column].dropna().str.split(',').explode().str.strip().value_counts()

fav_sem1 = count_favorites('mk_suka_sem1').rename('semester_1')
fav_sem2 = count_favorites('mk_suka_sem2').rename('semester_2')
fav_sem3 = count_favorites('mk_suka_sem3').rename('semester_3')
fav_sem4 = count_favorites('mk_suka_sem4').rename('semester_4')

# Gabungkan jadi satu dataframe
matkul_fav_df = pd.concat([fav_sem1, fav_sem2, fav_sem3, fav_sem4], axis=1).fillna(0).astype(int)

# Hitung organisasi
organisasi_counts = df['organisasi'].dropna().str.split(',').explode().str.strip().value_counts()
organisasi_df = organisasi_counts.rename('jumlah_mahasiswa').to_frame()

# Gabungkan semua hasil ke satu sheet menggunakan Excel
output_excel_path = "hasil_analisis_performa.xlsx"
with pd.ExcelWriter(output_excel_path) as writer:
    df.to_excel(writer, sheet_name="Data_Bersih", index=False)
    matkul_fav_df.to_excel(writer, sheet_name="Matkul_Favorit")
    organisasi_df.to_excel(writer, sheet_name="Organisasi")

output_excel_path
```
